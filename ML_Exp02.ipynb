```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Data Collection
df =pd.read_csv("student_scores.csv")

# Independent and dependent variables
x = df["Hours"]
y = df["Scores"]

# Number of observations
n = len(df)

# Required summations
sum_x = x.sum()
sum_y = y.sum()
sum_xy = (x * y).sum()
sum_x2 = (x ** 2).sum()

# Calculate slope (m)
m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)

# Calculate intercept (c)
c = (sum_y - m * sum_x) / n

print("Slope (m):", m)
print("Intercept (c):", c)

# Predict values
y_pred = m * x + c

# Display actual and predicted values
result = pd.DataFrame({
    "Hours": x,
    "Actual": y,
    "Predicted": y_pred
})

print(result)

plt.scatter(x, y, color="blue", label="Actual")
plt.plot(x, y_pred, color="red", label="Predicted")
plt.xlabel("Hours")
plt.ylabel("Scores")
plt.legend()
plt.show()
```
