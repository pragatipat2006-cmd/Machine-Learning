# Simple Decision Tree Implementation
### Without Scikit-learn

This notebook builds a small decision tree using **Python, pandas, and entropy/information gain**.

## 1. Import libraries

```python
import pandas as pd
import math
from collections import Counter
```

## 2. Create a small dataset

```python
data = {
    'Weather': ['Sunny', 'Sunny', 'Rainy', 'Rainy', 'Sunny', 'Rainy'],
    'Temperature': ['Hot', 'Cool', 'Hot', 'Cool', 'Hot', 'Cool'],
    'Play': ['No', 'Yes', 'Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
print(df)
```

## 3. Calculate entropy

Entropy measures how mixed the target labels are. A lower entropy means less uncertainty.

```python
def entropy(target):
    counts = Counter(target)
    total = len(target)
    result = 0

    for count in counts.values():
        p = count / total
        result -= p * math.log2(p)

    return result
```

## 4. Calculate information gain

```python
def information_gain(df, feature, target):
    total_entropy = entropy(df[target])
    weighted_entropy = 0

    for value in df[feature].unique():
        subset = df[df[feature] == value]
        weighted_entropy += (len(subset) / len(df)) * entropy(subset[target])

    return total_entropy - weighted_entropy
```

## 5. Build the decision tree

The feature with the highest information gain is selected at each step.

```python
def build_tree(df, features, target):
    # Stop if all rows have the same target label
    if len(df[target].unique()) == 1:
        return df[target].iloc[0]

    # Stop if no features remain
    if len(features) == 0:
        return df[target].mode()[0]

    # Choose the feature with the highest information gain
    best_feature = max(
        features,
        key=lambda f: information_gain(df, f, target)
    )

    tree = {best_feature: {}}

    for value in df[best_feature].unique():
        subset = df[df[best_feature] == value]
        remaining_features = [f for f in features if f != best_feature]
        tree[best_feature][value] = build_tree(
            subset, remaining_features, target
        )

    return tree
```

## 6. Train and display the decision tree

```python
features = ['Weather', 'Temperature']
tree = build_tree(df, features, 'Play')

print('Decision Tree:')
print(tree)
```

## 7. Interpret the result

- If `Temperature` is `Cool`, predict `Yes`.
- If `Temperature` is `Hot` and `Weather` is `Sunny`, predict `No`.
- If `Temperature` is `Hot` and `Weather` is `Rainy`, predict `Yes`.

### Key points
1. **Entropy** measures uncertainty.
2. **Information gain** measures how much a feature reduces uncertainty.
3. The tree chooses the feature with the highest information gain.
4. This is a simple educational implementation for categorical data; it does not use Scikit-learn.
